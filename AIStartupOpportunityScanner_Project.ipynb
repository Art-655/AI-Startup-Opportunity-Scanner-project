# AI Startup Opportunity Scanner

This notebook runs the core data, scoring, and machine-learning workflow using the processed project datasets. The Streamlit UI is launched separately with `py -3 -m streamlit run dashboard/app.py`.

````python
from pathlib import Path
import pandas as pd

from src.data_processing import DataCleaner
from src.ml_models import SuccessPredictor
from src.opportunity_engine import OpportunityScorer

PROJECT_ROOT = Path.cwd()
PROCESSED = PROJECT_ROOT / 'data' / 'processed'
````

## Load Processed Data
The cleaning script creates these files from the attached startup datasets.

````python
ai_funding = pd.read_csv(PROCESSED / 'ai_funding.csv')
indian_funding = pd.read_csv(PROCESSED / 'indian_funding.csv')
producthunt = pd.read_csv(PROCESSED / 'producthunt.csv')
startup_success = pd.read_csv(PROCESSED / 'startup_success.csv')

{
    'ai_funding_rows': len(ai_funding),
    'indian_funding_rows': len(indian_funding),
    'producthunt_rows': len(producthunt),
    'startup_success_rows': len(startup_success),
    'outcomes': startup_success['outcome'].value_counts().to_dict(),
}
````

## Build Opportunity Scores
The dashboard derives category-level demand, funding, competition, momentum, and failure-risk features from the real processed records.

````python
ai = ai_funding.copy()
indian = indian_funding.copy()
ai['sector'] = ai['sector'].fillna('Unknown')
indian['sector'] = indian['Industry'].fillna('Unknown')
ai['amount_usd'] = ai['amount_usd_millions'].fillna(0) * 1_000_000
indian['amount_usd'] = indian['Funding_Amount_USD'].fillna(0)
funding = pd.concat([
    ai[['sector', 'amount_usd', 'year']],
    indian[['sector', 'amount_usd', 'Funding_Year']].rename(columns={'Funding_Year': 'year'}),
], ignore_index=True)
funding_summary = funding.groupby('sector', as_index=False).agg(
    funding_growth=('amount_usd', 'sum'),
    competition_level=('sector', 'size'),
)
trend_summary = funding.groupby('sector', as_index=False).agg(
    trend_momentum=('year', lambda values: float((values >= values.max() - 2).sum()))
)
outcome_summary = startup_success.groupby('sector', as_index=False).agg(
    market_demand=('market_size_billion', 'median'),
    failure_risk=('outcome', lambda values: (values == 'Failure').mean()),
)
features = funding_summary.merge(trend_summary, on='sector').merge(outcome_summary, on='sector', how='left')
features['market_demand'] = features['market_demand'].fillna(startup_success['market_size_billion'].median())
features['failure_risk'] = features['failure_risk'].fillna((startup_success['outcome'] == 'Failure').mean())
scored_categories = OpportunityScorer().calculate_scores(features)
scored_categories[['sector', 'opportunity_score']].head(10)
````

## Train or Reload the Success Model
The outcome is normalized at the model boundary: `Failure` is 0 and other observed outcomes are 1.

````python
features = startup_success.drop(columns=['outcome'])
target = startup_success['outcome']
predictor = SuccessPredictor(
    model_path=PROJECT_ROOT / 'models' / 'success_predictor.joblib',
    n_estimators=30,
    max_depth=4,
    learning_rate=0.1,
)
predictor.train(features, target)
predictor.predict_probability(features.head())
````
